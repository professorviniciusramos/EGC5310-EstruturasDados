# Preparação para os experimentos

````python
import time
import statistics
import pandas as pd
import matplotlib.pyplot as plt
````

Usaremos uma função simples para repetir medições e calcular a mediana.

A mediana ajuda a reduzir o efeito de execuções ocasionalmente mais lentas.

````python
def medir(funcao, repeticoes=7):
    tempos = []

    for _ in range(repeticoes):
        inicio = time.perf_counter()
        funcao()
        fim = time.perf_counter()
        tempos.append(fim - inicio)

    return statistics.median(tempos)
````

## 2.1 Hipótese — inserção

Quando `n` aumentar, qual operação deverá apresentar maior crescimento de tempo?

- `append()`
- `insert(0, ...)`

**Justificativa:**

> Escreva aqui.

````python
tamanhos = [1_000, 10_000, 100_000, 500_000]
resultados_insercao = []

for n in tamanhos:
    base = list(range(n))

    def testar_append():
        lista = base.copy()
        lista.append(-1)

    def testar_insert_inicio():
        lista = base.copy()
        lista.insert(0, -1)

    tempo_append = medir(testar_append)
    tempo_insert = medir(testar_insert_inicio)

    resultados_insercao.append({
        "n": n,
        "append": tempo_append,
        "insert_inicio": tempo_insert
    })

df_insercao = pd.DataFrame(resultados_insercao)
df_insercao
````

````python
plt.figure(figsize=(8, 5))
plt.plot(df_insercao["n"], df_insercao["append"], marker="o", label="append")
plt.plot(df_insercao["n"], df_insercao["insert_inicio"], marker="o", label="insert(0, ...)")
plt.xlabel("n")
plt.ylabel("tempo (s)")
plt.title("Inserção em lista")
plt.legend()
plt.grid(True)
plt.show()
````

# Atividade complementar — Sequência de operações

Considere duas estratégias para incluir novos alunos em um ranking.

````python
def gerar_alunos(n):
    return [
        {"nome": f"Aluno {i}", "nota": (i * 7) % 101 / 10}
        for i in range(n)
    ]
````

## Estratégia A

```python
for aluno in novos_alunos:
    ranking.append(aluno)
    ranking.sort(key=lambda x: x["nota"], reverse=True)
```

## Estratégia B

```python
for aluno in novos_alunos:
    ranking.append(aluno)

ranking.sort(key=lambda x: x["nota"], reverse=True)
```

Antes de executar:

> Qual deve ser mais eficiente quando a quantidade de novos alunos crescer? Por quê?

````python
def estrategia_a(n):
    ranking = []
    novos_alunos = gerar_alunos(n)

    for aluno in novos_alunos:
        ranking.append(aluno)
        ranking.sort(key=lambda x: x["nota"], reverse=True)

    return ranking


def estrategia_b(n):
    ranking = []
    novos_alunos = gerar_alunos(n)

    for aluno in novos_alunos:
        ranking.append(aluno)

    ranking.sort(key=lambda x: x["nota"], reverse=True)
    return ranking
````

````python
tamanhos_ranking = [100, 500, 1_000, 2_000]

resultados_ranking = []

for n in tamanhos_ranking:
    tempo_a = medir(lambda: estrategia_a(n), repeticoes=3)
    tempo_b = medir(lambda: estrategia_b(n), repeticoes=3)

    resultados_ranking.append({
        "n": n,
        "ordenar_a_cada_insercao": tempo_a,
        "ordenar_uma_vez": tempo_b
    })

df_ranking = pd.DataFrame(resultados_ranking)
df_ranking
````

# Atividade 3 — DataFrame: registro a registro × lote

Agora vamos investigar uma situação comum em Ciência de Dados.

Queremos criar um DataFrame a partir de muitos registros.

## 3.1 Hipótese

Qual estratégia você espera que seja melhor?

### Estratégia A — concatenar um registro por vez

```python
for registro in dados:
    novo = pd.DataFrame([registro])
    df = pd.concat([df, novo], ignore_index=True)
```

### Estratégia B — criar em lote

```python
df = pd.DataFrame(dados)
```

**Sua previsão:**

> Escreva aqui.

**Por quê?**

> Escreva aqui.

````python
def gerar_registros(n):
    return [
        {
            "matricula": i,
            "curso": "CD" if i % 2 == 0 else "CC",
            "nota": (i * 13) % 101 / 10
        }
        for i in range(n)
    ]
````

````python
def dataframe_incremental(registros):
    df = pd.DataFrame(columns=["matricula", "curso", "nota"])

    for registro in registros:
        novo = pd.DataFrame([registro])
        df = pd.concat([df, novo], ignore_index=True)

    return df


def dataframe_em_lote(registros):
    return pd.DataFrame(registros)
````

````python
tamanhos_df = [100, 500, 1_000, 2_000]
resultados_df = []

for n in tamanhos_df:
    registros = gerar_registros(n)

    tempo_incremental = medir(
        lambda: dataframe_incremental(registros),
        repeticoes=3
    )

    tempo_lote = medir(
        lambda: dataframe_em_lote(registros),
        repeticoes=3
    )

    resultados_df.append({
        "n": n,
        "incremental": tempo_incremental,
        "lote": tempo_lote
    })

df_benchmark = pd.DataFrame(resultados_df)
df_benchmark
````

````python
plt.figure(figsize=(8, 5))
plt.plot(df_benchmark["n"], df_benchmark["incremental"], marker="o", label="incremental")
plt.plot(df_benchmark["n"], df_benchmark["lote"], marker="o", label="lote")
plt.xlabel("n")
plt.ylabel("tempo (s)")
plt.title("Construção de DataFrame")
plt.legend()
plt.grid(True)
plt.show()
````
