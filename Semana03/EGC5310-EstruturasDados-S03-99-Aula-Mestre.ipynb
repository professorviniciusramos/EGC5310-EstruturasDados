## Primeiro: uma lista tem posições

```text
índice      0       1       2       3
         ┌───────┬───────┬───────┬───────┐
lista →  │ Ana   │ Bruno │ Carla │ Diego │
         └───────┴───────┴───────┴───────┘
```

::: {.fragment .question-big}

````python
alunos = [
    {"matricula": 101, "nome": "Ana"},
    {"matricula": 102, "nome": "Bruno"},
    {"matricula": 103, "nome": "Carla"},
    {"matricula": 104, "nome": "Diego"},
]
alunos[2]
````

<div class="concept-definition">`lista[i]` → `O(1)`</div>

## Inserir no final


<div class="fragment small-note">

</div>

## E se inserirmos no início?

```python
alunos.insert(0, novo_aluno)
```

```text
antes
[A][B][C][D]

        deslocamentos →
[ ][A][B][C][D]

depois
[X][A][B][C][D]
```

````python
exemplo = ["A", "B", "C", "D"]
exemplo.insert(0, "X")
exemplo
````

## A posição mudou o trabalho {.concept-slide}

<div class="concept-definition">`insert(0, x)` → `O(n)`</div>

::: {.fragment .question-big}

:::

## Uma linha de código ≠ uma quantidade de trabalho {.question-slide}

```python
alunos.append(x)
```

```python
alunos.insert(0, x)
```

<div class="fragment prompt-gold">
As duas ocupam uma linha. Os mecanismos são diferentes.
</div>

## E para remover?

:::: {.columns}
::: {.column width="48%"}
### Final
```python
lista.pop()
```
:::
::: {.column width="48%"}
### Início
```python
lista.pop(0)
```
:::
::::

::: {.fragment .question-big}
Qual delas pode exigir deslocamentos?
:::

````python
lista = list(range(10))
ultimo = lista.pop()
primeiro = lista.pop(0)
print("removido do final:", ultimo)
print("removido do início:", primeiro)
print("lista restante:", lista)
````

## O que já conseguimos explicar?

| Operação | Comportamento | Mecanismo |
|---|---:|---|
| `lista[i]` | `O(1)` | acesso por posição |
| `append(x)` | `O(1)` amortizado | final + crescimento ocasional |
| `insert(0,x)` | `O(n)` | deslocamento |
| `pop()` | `O(1)` | final |
| `pop(0)` | `O(n)` | deslocamento |

Chega um novo estudante e precisamos manter um **ranking por nota**.

````python
ranking = [
    {"nome": "Ana", "nota": 9.5},
    {"nome": "Bruno", "nota": 8.7},
    {"nome": "Carla", "nota": 8.2},
]
novo_aluno = {"nome": "Diego", "nota": 9.1}
ranking.append(novo_aluno)
ranking
````

```text
       ↓
```

## Como informar o critério de ordenação?

Primeiro, uma função comum:

```python
def obter_nota(aluno):
    return aluno["nota"]
```

Depois:

```python
ranking.sort(
    key=obter_nota,
    reverse=True
)
```

<div class="fragment small-note"><code>key=</code> informa qual valor deve orientar a ordenação.</div>

````python
def obter_nota(aluno):
    return aluno["nota"]

ranking.sort(key=obter_nota, reverse=True)
ranking
````

## Uma pequena função sem nome

```python
def obter_nota(aluno):
    return aluno["nota"]
```

pode ser escrita como:

```python
lambda aluno: aluno["nota"]
```

```text
lambda aluno: aluno["nota"]
       ↑             ↑
    entrada        saída
```

::: {.fragment .prompt-gold}
`lambda` não muda a ideia: recebe um aluno e devolve sua nota.
:::

````python
ranking.sort(
    key=lambda aluno: aluno["nota"],
    reverse=True
)
ranking
````

## “Professor, eu usaria um <span class="gold">DataFrame</span>” {.question-slide}

<div class="fragment question-big">Por que todo mundo pensa em DataFrame em Ciência de Dados?</div>

````python
import pandas as pd

dados = [
    {"nome": "Ana", "curso": "CD", "nota": 8.5},
    {"nome": "Bruno", "curso": "CC", "nota": 7.2},
    {"nome": "Carla", "curso": "CD", "nota": 9.1},
]

df = pd.DataFrame(dados)
df
````

## Uma abstração orientada a dados tabulares

```text
             variáveis
          ↓       ↓       ↓
       nome     curso    nota
      ┌────────┬───────┬──────┐
linha │ Ana    │ CD    │ 8.5  │
linha │ Bruno  │ CC    │ 7.2  │
linha │ Carla  │ CD    │ 9.1  │
      └────────┴───────┴──────┘
```

<div class="fragment prompt-gold">As operações se aproximam da linguagem da análise de dados.</div>

````python
print("média:", df["nota"].mean())
print("\naprovados:")
display(df[df["nota"] >= 7])
print("\nmédia por curso:")
display(df.groupby("curso")["nota"].mean())
````

## Por isso DataFrame aparece tanto


<div class="fragment small-note">
Um comportamento quadrático pode surgir de uma decisão plausível de programação.

## A pergunta está incompleta

<div class="concept-definition">JSON é um formato de **representação e intercâmbio**.</div>

<div class="fragment small-note">
Ele não ocupa exatamente o mesmo papel que uma `list` ou um `DataFrame` em memória.
</div>

````python
import json

texto_json = '''
[
  {"matricula": 101, "nome": "Ana"},
  {"matricula": 102, "nome": "Bruno"}
]
'''

dados_json = json.loads(texto_json)

print(type(dados_json))
print(type(dados_json[0]))
print(dados_json)
````

## Depois do parsing...

```text
JSON
 ↓
json.loads(...)
 ↓
list
 └── dict
      ├── str
      ├── int
      └── ...
```

::: {.fragment .prompt-gold}
As operações de manipulação passam a ocorrer sobre estruturas em memória.
:::

## O trabalho escondido no JSON

```text
arquivo / mensagem JSON
          ↓
        leitura
          ↓
        parsing
          ↓
  estruturas em memória
          ↓
      manipulação
          ↓
     serialização
          ↓
        escrita
```

<div class="fragment small-note">
Para um documento de tamanho `m`, o parsing completo pode ser modelado introdutoriamente como aproximadamente `O(m)`.
</div>

## Um pipeline real pode usar os três

```text
API
 ↓
JSON
 ↓ desserialização
list / dict
 ↓ manipulação incremental
DataFrame
 ↓ análise tabular
JSON
 ↓ serialização
outro sistema
```

::: {.fragment .prompt-gold}
Não existe um vencedor absoluto.
:::

## Então quando usar cada um?

| Necessidade | `list` | `DataFrame` | JSON |
|---|---|---|---|
| crescer registro a registro | muito natural | geralmente não ideal | não é a finalidade |
| manipular objetos Python | natural | possível | exige parsing |
| operações sobre colunas | manual | natural | não é a finalidade |
| filtros/agregações tabulares | manual | natural | exige processamento |
| intercâmbio entre sistemas | Python-específico | exige serialização | muito apropriado |

## O repertório cresceu

| Operação / situação | Modelo |
|---|---:|
| `lista[i]` | `O(1)` |
| `append()` | `O(1)` amortizado |

| operação sobre `n` valores | `O(n)` |
| ordenação eficiente | `O(n log n)` típico |
| concatenações crescentes repetidas | pode aproximar `O(n²)` |
| parsing completo de JSON | aproximadamente `O(m)` |
