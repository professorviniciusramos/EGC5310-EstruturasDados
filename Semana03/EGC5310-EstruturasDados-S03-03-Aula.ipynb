# Abrindo a caixa-preta da `list`

````python
alunos = [
    {"matricula": 101, "nome": "Ana"},
    {"matricula": 102, "nome": "Bruno"},
    {"matricula": 103, "nome": "Carla"},
    {"matricula": 104, "nome": "Diego"},
]
alunos[2]
````

## Acesso por posição

```text
índice      0       1       2       3
         ┌───────┬───────┬───────┬───────┐
lista →  │ Ana   │ Bruno │ Carla │ Diego │
         └───────┴───────┴───────┴───────┘
```

`alunos[2]` não exige testar os elementos anteriores.

> **Modelo:** `O(1)`

````python
novo_aluno = {"matricula": 105, "nome": "Eva"}
alunos.append(novo_aluno)
alunos
````

## Inserir no final

Uma lista dinâmica mantém espaço para crescer.

```text
tamanho = 3
capacidade = 5

[A][B][C][ ][ ]
```

Quando existe espaço, adicionar ao final é barato.

## E quando o espaço termina?

```text
[A][B][C][D]
```

Chega `E`.

Simplificando:
1. obter espaço maior;
2. transferir referências;
3. inserir o novo elemento.

Nem todo `append()` individual realiza exatamente o mesmo trabalho.

## Custo amortizado

A maioria dos `append()` é barata; ocasionalmente a estrutura precisa crescer.

> **`append()` é `O(1)` amortizado.**

Guardaremos o termo sem fazer a demonstração formal nesta semana.

````python
exemplo = ["A", "B", "C", "D"]
exemplo.insert(0, "X")
exemplo
````

## Inserir no início

```text
[A][B][C][D]

 →  →  →  →
[ ][A][B][C][D]

[X][A][B][C][D]
```

Se há `n` elementos, muitos podem precisar mudar de posição.

> **Modelo:** `O(n)`

# Uma linha de código ≠ uma quantidade de trabalho

```python
alunos.append(x)
```

e

```python
alunos.insert(0, x)
```

ocupam uma linha, mas escondem mecanismos diferentes.

````python
lista = list(range(10))
ultimo = lista.pop()
primeiro = lista.pop(0)
ultimo, primeiro, lista
````

## Remover também depende da posição

- `pop()` remove do final.
- `pop(0)` remove do início e pode exigir deslocamentos.

## Primeira síntese

| Operação | Comportamento | Mecanismo |
|---|---:|---|
| `lista[i]` | `O(1)` | acesso por posição |
| `append(x)` | `O(1)` amortizado | final + crescimento ocasional |
| `insert(0,x)` | `O(n)` | deslocamento |
| `pop()` | `O(1)` | remoção no final |
| `pop(0)` | `O(n)` | deslocamento |

# Operações formam processos

````python
ranking = [
    {"nome": "Ana", "nota": 9.5},
    {"nome": "Bruno", "nota": 8.7},
    {"nome": "Carla", "nota": 8.2},
]
novo_aluno = {"nome": "Diego", "nota": 9.1}
ranking.append(novo_aluno)
ranking
````

## Inserir é apenas parte do problema

Depois da inserção, queremos manter o ranking ordenado.

```python
ranking.append(novo_aluno)
ranking.sort(...)
```

Ordenar também é trabalho.

## Ordenar também é uma operação

```text
dados desordenados
       ↓
algoritmo de ordenação
       ↓
dados ordenados
```

Sem estudar algoritmos de ordenação agora, guardamos que algoritmos eficientes costumam apresentar comportamento da ordem de `O(n log n)`.

````python
def obter_nota(aluno):
    return aluno["nota"]

ranking.sort(key=obter_nota, reverse=True)
ranking
````

## Como o Python sabe o critério?

```python
def obter_nota(aluno):
    return aluno["nota"]
```

`key=` recebe uma função que devolve o valor usado na ordenação.

## Uma função pequena sem nome

```python
def obter_nota(aluno):
    return aluno["nota"]
```

pode ser escrita como:

```python
lambda aluno: aluno["nota"]
```

```text
lambda aluno: aluno["nota"]
       ↑             ↑
    entrada        saída
```

Uma `lambda` é uma pequena função sem nome.

````python
ranking.sort(
    key=lambda aluno: aluno["nota"],
    reverse=True
)
ranking
````

## Duas estratégias corretas

**A — ordenar após cada inserção**

```python
for aluno in novos_alunos:
    ranking.append(aluno)
    ranking.sort(key=lambda x: x["nota"], reverse=True)
```

**B — ordenar uma vez**

```python
for aluno in novos_alunos:
    ranking.append(aluno)

ranking.sort(key=lambda x: x["nota"], reverse=True)
```

Mesmo objetivo final. Mesma quantidade de trabalho?

# “Professor, eu usaria um DataFrame”

````python
import pandas as pd

dados = [
    {"nome": "Ana", "curso": "CD", "nota": 8.5},
    {"nome": "Bruno", "curso": "CC", "nota": 7.2},
    {"nome": "Carla", "curso": "CD", "nota": 9.1},
]

df = pd.DataFrame(dados)
df
````

## Por que DataFrame é tão importante?

Ele oferece uma abstração orientada a dados tabulares e operações muito próximas das perguntas analíticas.

````python
media = df["nota"].mean()
aprovados = df[df["nota"] >= 7]
media_por_curso = df.groupby("curso")["nota"].mean()

media, aprovados, media_por_curso
````

## Operações tabulares aparecem naturalmente

```python
df["nota"].mean()
df[df["nota"] >= 7]
df.groupby("curso")["nota"].mean()
df.sort_values("nota")
```

Isso ajuda a explicar por que DataFrames aparecem tanto em Ciência de Dados.

## List × DataFrame: ambos podem ser `O(n)`

Lista:

```python
for aluno in alunos:
    aluno["nota"] += 1
```

DataFrame:

```python
df["nota"] += 1
```

Nos dois casos, `n` valores são processados.

> Em um modelo simples, ambos são `O(n)`.

## Se ambos são `O(n)`, por que os tempos diferem?

```text
Lista Python      T(n) ≈ a · n
DataFrame         T(n) ≈ b · n
```

As constantes podem ser muito diferentes.

Motivos possíveis:
- overhead do interpretador;
- representação de memória;
- código compilado;
- vetorização;
- processamento em lote.

> **Mesma classe assintótica não significa mesmo tempo.**

## DataFrame é sempre melhor?

Não.

Uma lista cresce naturalmente:

```python
alunos.append(novo_aluno)
```

Adicionar repetidamente uma linha a um DataFrame pode exigir reconstruções/cópias e ser uma estratégia ruim.

````python
novo_aluno = {"nome": "Diego", "curso": "CD", "nota": 8.9}
novo_df = pd.DataFrame([novo_aluno])
df2 = pd.concat([df, novo_df], ignore_index=True)
df2
````

## Registro por registro × lote

**Incremental**

```python
for aluno in novos_alunos:
    novo_df = pd.DataFrame([aluno])
    df = pd.concat([df, novo_df], ignore_index=True)
```

**Em lote**

```python
novos_df = pd.DataFrame(novos_alunos)
df = pd.concat([df, novos_df], ignore_index=True)
```

> DataFrames são particularmente fortes quando operamos sobre conjuntos de dados.

## De onde pode surgir `O(n²)`?

Se concatenamos repetidamente estruturas cada vez maiores:

```text
1 + 2 + 3 + ... + n
```

o trabalho acumulado pode crescer quadraticamente.

> **`O(n²)`**

É um exemplo de comportamento ruim produzido por uma decisão plausível de programação.

# “E JSON?”

## JSON é uma representação de dados

```json
[
  {"matricula": 101, "nome": "Ana"},
  {"matricula": 102, "nome": "Bruno"}
]
```

> **Qual é o Big-O do JSON?**

A pergunta, isoladamente, está mal formulada.

JSON é um formato de representação/intercâmbio.

````python
import json

texto_json = '''
[
  {"matricula": 101, "nome": "Ana"},
  {"matricula": 102, "nome": "Bruno"}
]
'''

dados_json = json.loads(texto_json)
type(dados_json), type(dados_json[0]), dados_json
````

## Depois do parsing, voltamos a estruturas em memória

```text
JSON
 ↓
json.loads(...)
 ↓
list
 └── dict
      ├── str
      ├── int
      └── ...
```

As operações passam a ocorrer sobre essas estruturas.

## O trabalho escondido no JSON

```text
arquivo / mensagem JSON
          ↓
        leitura
          ↓
        parsing
          ↓
  estruturas em memória
          ↓
      manipulação
          ↓
     serialização
          ↓
        escrita
```

Para um documento textual de tamanho `m`, o parsing completo é, em um modelo introdutório, aproximadamente `O(m)`.

# Um pipeline real pode usar os três

```text
API
 ↓
JSON
 ↓ desserialização
list / dict
 ↓ manipulação incremental
DataFrame
 ↓ análise tabular
JSON
 ↓ serialização
outro sistema
```

> **Não existe um vencedor absoluto. Cada representação pode cumprir uma função diferente.**

## Então quando usar cada um?

| Necessidade | `list` | `DataFrame` | JSON |
|---|---|---|---|
| crescer registro a registro | muito natural | geralmente não é o uso ideal | não é a finalidade |
| manipular objetos Python | natural | possível | exige parsing |
| operações sobre colunas | manual | natural | não é a finalidade |
| filtros/agregações tabulares | manual | natural | exige processamento |
| agrupamento | explícito | nativo | não é a finalidade |
| intercâmbio entre sistemas | Python-específico | exige serialização | muito apropriado |
| estrutura hierárquica serializada | não é formato | não é formato | muito apropriado |

# Qual é o Big-O da estrutura?

Perguntas como “qual é o Big-O de uma lista/DataFrame/JSON?” são incompletas.

> **A pergunta correta é: qual é a complexidade desta operação sobre esta representação?**

## Síntese da Semana 03

| Operação / situação | Modelo |
|---|---:|
| `lista[i]` | `O(1)` |
| `list.append()` | `O(1)` amortizado |
| `list.insert(0,x)` | `O(n)` |
| `list.pop()` | `O(1)` |
| `list.pop(0)` | `O(n)` |
| percurso de `n` registros | `O(n)` |
| operação vetorizada sobre `n` valores | `O(n)` em modelo simples |
| ordenação eficiente | `O(n log n)` típico |
| concatenações crescentes repetidas | pode aproximar `O(n²)` |
| parsing completo de JSON | aproximadamente `O(m)` |

# O princípio da semana

> **Não existe “o custo de uma estrutura”.**

Existe o custo de:
1. uma **operação**;
2. sobre uma **representação**;
3. dentro de uma **sequência de operações**;
4. para um determinado **tamanho de entrada**.

O tempo medido ainda depende da implementação concreta e do ambiente.
