## E se organizássemos os dados de outra maneira?

```text
              1082
             /    \
          1047    1201
          /       /  \
       1023    1125  1350
```

::: {.fragment .question-big}
Para procurar 1125, precisamos examinar todos os estudantes?
:::

<div class="fragment prompt-gold">1082 → 1201 → 1125</div>

````python
class No:
    def __init__(self, matricula, estudante):
        self.matricula=matricula; self.estudante=estudante; self.esquerda=None; self.direita=None

def inserir(no, matricula, estudante):
    if no is None: return No(matricula, estudante)
    if matricula < no.matricula: no.esquerda=inserir(no.esquerda, matricula, estudante)
    else: no.direita=inserir(no.direita, matricula, estudante)
    return no

raiz=None
for matricula,nome in [(1082,"Carla"),(1047,"Bruno"),(1201,"Eva"),(1023,"Ana"),(1125,"Daniel"),(1350,"Fábio")]:
    raiz=inserir(raiz, matricula, {"nome":nome})
````

## Podemos entender uma árvore sem saber implementá-la?

```python
def buscar_arvore(no, matricula):
    if no is None:
        return None
    if matricula == no.matricula:
        return no.estudante
    if matricula < no.matricula:
        return buscar_arvore(no.esquerda, matricula)
    return buscar_arvore(no.direita, matricula)
```

::: {.fragment .prompt-gold}
**Você não precisa saber produzir esse código hoje.** Consegue entender a estratégia?
:::

````python
def buscar_arvore_com_caminho(no, matricula, caminho=None):
    if caminho is None: caminho=[]
    if no is None: return None, caminho
    caminho=caminho+[no.matricula]
    if matricula == no.matricula: return no.estudante, caminho
    if matricula < no.matricula: return buscar_arvore_com_caminho(no.esquerda, matricula, caminho)
    return buscar_arvore_com_caminho(no.direita, matricula, caminho)
resultado,caminho=buscar_arvore_com_caminho(raiz,1125)
print("caminho:"," → ".join(map(str,caminho)))
print("resultado:",resultado)
````
