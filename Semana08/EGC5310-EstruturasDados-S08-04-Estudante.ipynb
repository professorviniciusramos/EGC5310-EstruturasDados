Antes do código, imagine `[3, 7, 8]` como a parte ordenada e o próximo valor `1`. Quais elementos deslocam? Onde o 1 entra?

Complete somente dois trechos da função fornecida:

- **A02_DESLOCAR:** valor que deve ocupar `valores[j + 1]` durante o deslocamento.
- **A02_INSERIR:** valor que deve ocupar a posição final aberta.

`atual` guarda o elemento a inserir. `j` percorre a parte ordenada da direita para a esquerda. A função modifica a lista recebida e também a retorna; use cópias nos experimentos.


### Registro A02 — Previsão

**Elementos deslocados e posição final de 1:** …

**Por que precisamos guardar atual?** …


```python
def insertion_sort(valores, mostrar_passos=False):
    for i in range(1, len(valores)):
        atual = valores[i]
        j = i - 1
        if mostrar_passos:
            print(f"i={i}, atual={atual}, prefixo={valores[:i]}")
        while j >= 0 and valores[j] > atual:
            valores[j + 1] = completar("A02_DESLOCAR")
            if mostrar_passos:
                print(f"  desloca {j} para {j + 1}: {valores}")
            j -= 1
        valores[j + 1] = completar("A02_INSERIR")
        if mostrar_passos:
            print(f"  insere em {j + 1}: {valores}")
    return valores
```

## Atividade 3 — Intercalação e Merge Sort (25–55 min)

Intercale `[1, 4, 7]` e `[2, 4, 6]`, que já estão ordenadas. Registre as quatro primeiras escolhas e de qual parte vem cada uma. Em empate, a implementação escolhe da esquerda.

Complete quatro trechos: avanço de `i`, avanço de `j`, sobra da esquerda e sobra da direita. Cada índice aponta o próximo elemento ainda não consumido.

A função recebe `key`, com padrão que devolve o próprio valor. Nesta atividade, trabalharemos com números. Na A7, você adaptará esse mecanismo para registros completos.


### Registro A03 — Previsão

**Quatro primeiras escolhas (valor e origem):** …

**Quando uma parte acabar, o que acontece com a outra?** …


```python
def intercalar(esquerda, direita, key=lambda x: x,
               mostrar_passos=False):
    resultado = []
    i = j = 0
    while i < len(esquerda) and j < len(direita):
        if key(esquerda[i]) <= key(direita[j]):
            escolhido = esquerda[i]
            i = completar("A03_AVANCAR_I")
            origem = "esquerda"
        else:
            escolhido = direita[j]
            j = completar("A03_AVANCAR_J")
            origem = "direita"
        resultado.append(escolhido)
        if mostrar_passos:
            print(f"  {origem}: {escolhido}, parcial={resultado}")
    resultado.extend(completar("A03_SOBRA_ESQUERDA"))
    resultado.extend(completar("A03_SOBRA_DIREITA"))
    return resultado
```

```python
resultado = intercalar([1, 4, 7], [2, 4, 6], mostrar_passos=True)
print("Resultado com sobras:", resultado)
assert resultado == sorted([1, 4, 7] + [2, 4, 6])
assert intercalar([], [2, 3]) == [2, 3]
assert intercalar([1, 2], []) == [1, 2]
```

### A recursão fornecida

Leia caso-base, divisão e retorno com o professor. A divisão por posição produz partes de tamanhos semelhantes. O algoritmo recebe suas soluções e as intercala.

Execute a célula da função abaixo depois de completar a intercalação. Limite os prints por profundidade para acompanhar somente os níveis de interesse.


```python
def merge_sort(valores, key=lambda x: x, mostrar_passos=False,
               nivel=0, limite_print=1):
    if len(valores) <= 1:
        return valores.copy()
    meio = len(valores) // 2
    ver = mostrar_passos and nivel <= limite_print
    if ver:
        print(f"nível {nivel}: {valores[:meio]} / {valores[meio:]}")
    esquerda = merge_sort(valores[:meio], key, mostrar_passos,
                          nivel + 1, limite_print)
    direita = merge_sort(valores[meio:], key, mostrar_passos,
                         nivel + 1, limite_print)
    resultado = intercalar(esquerda, direita, key)
    if ver:
        print(f"retorno nível {nivel}: {resultado}")
    return resultado
```

```python
resultado_merge = merge_sort(dados, mostrar_passos=True, limite_print=1)
assert resultado_merge == sorted(dados)
print("Saída:", resultado_merge)
```

## Atividade 4 — Pivô, partições e Quicksort (55–85 min)

Na sequência comum, compare pivôs 8 e 4. Preveja os grupos `menores`, `iguais` e `maiores`, preservando sua ordem de chegada.

Complete as condições **A04_MENORES** e **A04_MAIORES**. O ramo restante recebe valores iguais ao pivô. Depois execute o particionamento, ainda sem ordenar internamente os grupos.


### Registro A04 — Previsão

**Pivô 8 — grupos e tamanhos:** …

**Qual escolha parece mais equilibrada neste exemplo? Por quê?** …


```python
def particionar(valores, pivo, mostrar_passos=False):
    menores, iguais, maiores = [], [], []
    for valor in valores:
        if completar("A04_MENORES"):
            menores.append(valor)
        elif completar("A04_MAIORES"):
            maiores.append(valor)
        else:
            iguais.append(valor)
    if mostrar_passos:
        print("pivô:", pivo)
        print("menores:", menores)
        print("iguais:", iguais)
        print("maiores:", maiores)
        print("tamanhos:", len(menores), len(iguais), len(maiores))
    return menores, iguais, maiores
```

```python
for pivo in [8, 4]:
    grupos = particionar(dados, pivo, mostrar_passos=True)
    assert Counter(sum(grupos, [])) == Counter(dados)
print("Ocorrências preservadas nas partições.")
```

### Recursão nos grupos

A função abaixo está pronta. Ela usa a sua função `particionar`. O grupo `iguais` permanece entre as duas partes resolvidas. Para o pivô aleatório, fornecemos um gerador com semente.


```python
def quick_sort(valores, escolha="primeiro", rng=None,
               mostrar_passos=False, nivel=0, limite_print=1):
    if len(valores) <= 1:
        return valores.copy()
    if escolha == "primeiro":
        pivo = valores[0]
    elif escolha == "aleatorio":
        pivo = rng.choice(valores)
    else:
        raise ValueError("escolha deve ser primeiro ou aleatorio")
    ver = mostrar_passos and nivel <= limite_print
    if ver:
        print(f"nível {nivel}")
    menores, iguais, maiores = particionar(valores, pivo, ver)
    esquerda = quick_sort(menores, escolha, rng, mostrar_passos,
                          nivel + 1, limite_print)
    direita = quick_sort(maiores, escolha, rng, mostrar_passos,
                         nivel + 1, limite_print)
    resultado = esquerda + iguais + direita
    if ver:
        print(f"retorno nível {nivel}: {resultado}")
    return resultado
```

```python
entrada_ordenada = list(range(1, 9))
print("Primeiro pivô:")
quick_sort(entrada_ordenada, mostrar_passos=True, limite_print=1)
print("Pivô aleatório:")
quick_sort(entrada_ordenada, escolha="aleatorio", rng=random.Random(5310),
           mostrar_passos=True, limite_print=0)
assert quick_sort([4, 4, 4, 4]) == [4, 4, 4, 4]
assert quick_sort(dados) == sorted(dados)
```

## Atividade 5 — Counting Sort e domínio das chaves (85–100 min)

Dados sintéticos: `[3, 1, 3, 0, 5, 2, 1, 5]`, inteiros não negativos.

Preveja as frequências de 0 a 5. Complete **A05_CONTAR** com a nova frequência de um valor e **A05_RECONSTRUIR** com a lista de ocorrências a acrescentar à saída.

A inicialização usa `[0] * (max(valores) + 1)`. O índice da tabela representa o valor contado.


### Registro A05 — Previsão

**Frequências por valor:** …

**n e k neste exemplo:** …

**O que muda se incluirmos o valor 1.000.000?** …


```python
def counting_sort(valores, mostrar_passos=False):
    if not valores:
        return []
    if any(type(v) is not int or v < 0 for v in valores):
        raise ValueError("Use inteiros não negativos")
    contagens = [0] * (max(valores) + 1)
    for valor in valores:
        contagens[valor] = completar("A05_CONTAR")
    if mostrar_passos:
        print("frequências:", contagens)
    resultado = []
    for valor, frequencia in enumerate(contagens):
        resultado.extend(completar("A05_RECONSTRUIR"))
        if mostrar_passos and frequencia:
            print(f"valor={valor}, vezes={frequencia}: {resultado}")
    return resultado
```

```python
avaliacoes = [3, 1, 3, 0, 5, 2, 1, 5]
resultado_counting = counting_sort(avaliacoes, mostrar_passos=True)
assert resultado_counting == sorted(avaliacoes)
assert counting_sort([]) == []
print("Resultado conferido.")
```

### Condições do ensaio

Entradas sintéticas reproduzíveis. n = 100, 300, 600. Mediana de cinco repetições. Cópia preparada fora do cronômetro. Alocações internas do algoritmo permanecem dentro da medição. A validação da saída ocorre antes do cronômetro. As contagens instrumentadas de operações, quando usadas na discussão, pertencem a outra execução.

`sorted()` serve de referência prática otimizada. A diferença de tempo não depende apenas da complexidade. Não ampliar automaticamente o caso adverso nem aumentar o limite de recursão.


```python
def medir(algoritmo, entrada, repeticoes=5):
    esperado = sorted(entrada)
    assert algoritmo(entrada.copy()) == esperado
    tempos = []
    for _ in range(repeticoes):
        copia = entrada.copy()
        inicio = perf_counter()
        algoritmo(copia)
        tempos.append(perf_counter() - inicio)
    return median(tempos)

algoritmos = {
    "insertion": insertion_sort,
    "merge": merge_sort,
    "quick_primeiro": quick_sort,
    "sorted": sorted,
}
```

```python
tempos_ordem = []
for n in [100, 300, 600]:
    aleatoria = random.Random(5310 + n).sample(range(n), n)
    entradas = {"aleatória": aleatoria, "ordenada": list(range(n)),
                "invertida": list(range(n - 1, -1, -1))}
    for ordem, entrada in entradas.items():
        for nome, algoritmo in algoritmos.items():
            tempos_ordem.append({"n": n, "ordem": ordem,
                                  "algoritmo": nome,
                                  "tempo_ms": 1000 * medir(algoritmo, entrada)})
print("Ensaio concluído:", len(tempos_ordem), "condições")
```

```python
import matplotlib.pyplot as plt
fig, eixos = plt.subplots(1, 2, figsize=(10, 3.8))
for ax, nome in zip(eixos, ["insertion", "quick_primeiro"]):
    for ordem, cor in [("aleatória", "#0094D4"),
                       ("ordenada", "#CCAC71"), ("invertida", "#818285")]:
        linhas = [r for r in tempos_ordem
                  if r["algoritmo"] == nome and r["ordem"] == ordem]
        ax.plot([r["n"] for r in linhas], [r["tempo_ms"] for r in linhas],
                marker="o", label=ordem, color=cor)
    ax.set(title=nome, xlabel="Elementos (n)", ylabel="Mediana (ms)")
    ax.legend(frameon=False, fontsize=9)
    ax.grid(alpha=.2)
fig.suptitle("Entradas sintéticas, cinco repetições")
fig.tight_layout()
plt.show()
```

### Adaptação da intercalação para registros

Uma função `key` recebe o registro e devolve o valor usado na comparação. Complete:

- **A07_COMPARAR:** condição que compara as chaves dos dois registros atuais e escolhe da esquerda em empate.
- **A07_TRANSPORTAR:** objeto que deve entrar no resultado após a escolha.

Os índices e as sobras já estão corretos. A recursão fornecida abaixo propaga key às chamadas e à intercalação.


```python
def intercalar_registros(esquerda, direita, key=lambda x: x,
               mostrar_passos=False):
    resultado = []
    i = j = 0
    while i < len(esquerda) and j < len(direita):
        if completar("A07_COMPARAR"):
            escolhido = esquerda[i]
            i += 1
            origem = "esquerda"
        else:
            escolhido = direita[j]
            j += 1
            origem = "direita"
        resultado.append(completar("A07_TRANSPORTAR"))
        if mostrar_passos:
            print(f"  {origem}: {escolhido}, parcial={resultado}")
    resultado.extend(esquerda[i:])
    resultado.extend(direita[j:])
    return resultado
```

```python
def merge_sort_registros(valores, key=lambda x: x, mostrar_passos=False,
               nivel=0, limite_print=1):
    if len(valores) <= 1:
        return valores.copy()
    meio = len(valores) // 2
    ver = mostrar_passos and nivel <= limite_print
    if ver:
        print(f"nível {nivel}: {valores[:meio]} / {valores[meio:]}")
    esquerda = merge_sort_registros(valores[:meio], key, mostrar_passos,
                          nivel + 1, limite_print)
    direita = merge_sort_registros(valores[meio:], key, mostrar_passos,
                         nivel + 1, limite_print)
    resultado = intercalar_registros(esquerda, direita, key)
    if ver:
        print(f"retorno nível {nivel}: {resultado}")
    return resultado
```

```python
def obter_valor(registro):
    return registro["valor_item"]

# Comparações explícitas apenas no recorte de 8 registros.
pequenos = registros[:8]
esquerda = merge_sort_registros(pequenos[:4], key=obter_valor)
direita = merge_sort_registros(pequenos[4:], key=obter_valor)
print("Chaves esquerda:", [obter_valor(r) for r in esquerda])
print("Chaves direita:", [obter_valor(r) for r in direita])

def chave_com_print(registro):
    print(f"compara id={registro['id_linha']}: valor={registro['valor_item']:.2f}")
    return registro["valor_item"]

pequenos_ordenados = intercalar_registros(esquerda, direita, key=chave_com_print)
print("Depois da ordenação do recorte:")
mostrar_registros(pequenos_ordenados)
```

### Verificação de todos os 200 registros

Os ids verificam ocorrências, mas isoladamente não garantem a integridade dos atributos. A equivalência de registros completos com a referência complementa essa verificação.


```python
ordenados = merge_sort_registros(registros, key=obter_valor)
assert len(ordenados) == len(registros)
assert all(obter_valor(a) <= obter_valor(b)
           for a, b in zip(ordenados, ordenados[1:]))
assert Counter(r["id_linha"] for r in ordenados) == Counter(
    r["id_linha"] for r in registros)
assert ordenados == sorted(registros, key=obter_valor)
print("Ordem, quantidade, ids e registros completos conferidos.")
mostrar_registros(ordenados)
```
