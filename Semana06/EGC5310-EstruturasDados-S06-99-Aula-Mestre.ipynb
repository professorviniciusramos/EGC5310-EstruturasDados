## Chaves repetidas: repetição também é informação

No arquivo original, uma fatura pode aparecer em várias linhas:

```text
InvoiceNo    StockCode    Quantity
536365       85123A       6
536365       71053        6
536365       84406B       8
536366       22633        6
```

Aqui, `InvoiceNo` não identifica uma única linha.

**Uma venda → vários produtos.**


## Uma lista preserva as ocorrências

```python
vendas = [
    ("536365", "85123A", 6),
    ("536365", "71053", 6),
    ("536365", "84406B", 8),
    ("536366", "22633", 6),
]
```

A repetição permite responder perguntas como:
- quantas linhas existem na venda?
- quais produtos aparecem?
- qual a quantidade total?
- um produto apareceu mais de uma vez?

**Eliminar repetição pode significar eliminar informação.**


## O que acontece se usarmos set?

```python
produtos = ["A", "A", "B", "C"]
set(produtos)
# {"A", "B", "C"}
```

O `set` preserva **quais valores distintos existem**, mas não a multiplicidade.

Isso é ótimo quando queremos semântica de conjunto.

Não serve quando a quantidade de ocorrências é parte do problema.


## E um dict com chave repetida?

```python
vendas = {}
vendas["536365"] = "85123A"
vendas["536365"] = "71053"
```

Resultado:

```python
{"536365": "71053"}
```

A segunda atribuição **substitui** o valor anterior.

O problema não é o `dict`: é o modelo `chave → um único valor`.


## Uma estrutura pode conter outra estrutura

## Quando a semântica é realmente de conjunto

Agora queremos comparar **grupos de clientes**, sem contar quantas vezes cada cliente apareceu.

- `A`: clientes do Reino Unido
- `B`: clientes com alguma compra de quantidade ≥ 10

Nesse problema, duplicatas não acrescentam informação.

Por isso, `set` passa a representar diretamente a semântica desejada.


## Operações entre conjuntos em Python

```text
x in A   pertinência: x pertence a A?
A & B    interseção: elementos que estão em A e B
A | B    união: elementos que estão em A ou B (ou nos dois)
A - B    diferença: elementos de A que não estão em B
A ^ B    diferença simétrica: elementos que estão em apenas um dos conjuntos
```

Exemplo:

```text
A = {1, 2, 3}
B = {3, 4}

A & B = {3}
A | B = {1, 2, 3, 4}
A - B = {1, 2}
A ^ B = {1, 2, 4}
```

## List do CPython: referências
```text
array contíguo: [ref][ref][ref][ ][ ]
                    ↓    ↓    ↓
                 objA objB objC   (objetos em outros locais)
```

Tamanho lógico = 3; capacidade interna pode ser maior.


## Se a capacidade acaba
```text
ANTES   [A][B][C][D]              cheia
append(E) → redimensionar o bloco, ou alocar maior
DEPOIS  [A][B][C][D][E][ ][ ]     espaço futuro
```

Se houver realocação, referências podem ser copiadas. CPython reserva espaço extra; **não há regra de sempre dobrar**.


## Por que append é amortizado?

`barato → barato → caro (redimensiona) → barato → ...`

Uma chamada pode custar mais quando há redimensionamento. Em muitas inserções, esses custos se espaçam:

**O(1) amortizado por `append`.**

Remover metade dos elementos não implica reduzir imediatamente a capacidade pela metade.


## Inserção ordenada: encontrar não é inserir

Considere:

```text
[10][20][30][40][50]
```

Queremos inserir `25`.

1. `bisect` encontra rapidamente a posição entre `20` e `30`: **O(log n)**.
2. Mas a lista precisa abrir espaço:

```text
[10][20][  ][30][40][50]
          ↑
         25
```

Para isso, referências à direita precisam ser deslocadas.

**Localizar: O(log n) · Inserir na lista: O(n) · `insort`: O(n)**

## Uma alternativa comum em Ciência de Dados: NumPy

```python
# Python
resultado = []
for q, p in zip(quantidades, precos):
    resultado.append(q * p)

# NumPy
resultado = np.array(quantidades) * np.array(precos)
```

`np.array(...)` cria um objeto `numpy.ndarray`.

Nesta semana, **não vamos estudar a implementação interna do ndarray**. Vamos observar o comportamento da operação.


## Duas soluções podem continuar sendo O(n)

Tanto o laço Python quanto a operação com NumPy precisam processar `n` valores.

```text
Python + for   → O(n)
NumPy          → O(n)
```

Mas isso não significa que terão o mesmo tempo medido.

> **Big-O descreve como o custo cresce; não é um cronômetro.**


## Por que os tempos podem ser diferentes?

Big-O não descreve:
- quantos segundos uma operação leva;
- todas as constantes envolvidas;
- otimizações da implementação;
- custos de preparação.

NumPy pode executar operações sobre arrays usando implementações otimizadas, evitando que escrevamos e executemos o laço elemento a elemento em Python.

A conversão `np.array(lista)` também tem custo e deve ser considerada quando fizer parte do problema.
