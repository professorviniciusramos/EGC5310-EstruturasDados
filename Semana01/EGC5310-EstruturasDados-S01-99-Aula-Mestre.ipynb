## Um sistema acadêmico

Considere uma pequena coleção de estudantes:

| Matrícula | Nome | Curso |
|---:|---|---|
| 1023 | Ana | Ciência de Dados |
| 1047 | Bruno | Ciência de Dados |
| 1082 | Carla | Engenharia |
| 1091 | Diego | Ciência de Dados |

**Problema:** como localizar, pela matrícula, um estudante nessa coleção?

## Qual seria o seu algoritmo?

Queremos localizar a matrícula **1082**.

## Uma primeira solução

A ideia mais simples é percorrer a coleção e comparar cada matrícula com a matrícula procurada.

**Pergunta:** se a função produz a resposta correta, o problema está resolvido?

```python
# Preparação: somente biblioteca padrão do Python.
from time import perf_counter
from statistics import median

estudantes = [
    {"matricula": 1023, "nome": "Ana", "curso": "Ciência de Dados"},
    {"matricula": 1047, "nome": "Bruno", "curso": "Ciência de Dados"},
    {"matricula": 1082, "nome": "Carla", "curso": "Engenharia"},
    {"matricula": 1091, "nome": "Diego", "curso": "Ciência de Dados"},
]


def buscar(estudantes, matricula):
    for estudante in estudantes:
        if estudante["matricula"] == matricula:
            return estudante
    return None

buscar(estudantes, 1082)
```

### Parada de aprendizagem

Teste também:

- a primeira matrícula;
- a última matrícula;
- uma matrícula inexistente.

Pergunte-se: **o algoritmo realiza a mesma quantidade de trabalho em todos os casos?**

## Quanto trabalho foi necessário?

Considere a sequência:

**1023 → 1047 → 1082 → 1091**

Para encontrar `1082`, quantas comparações foram realizadas?

Em vez de começar pelo cronômetro, vamos observar uma operação do algoritmo que podemos contar diretamente.

```python
def buscar_contando(estudantes, matricula):
    comparacoes = 0

    for estudante in estudantes:
        comparacoes += 1
        if estudante["matricula"] == matricula:
            return estudante, comparacoes

    return None, comparacoes

buscar_contando(estudantes, 1082)
```

## O algoritmo realiza a mesma quantidade de trabalho?

Compare quatro situações:

- matrícula no primeiro registro;
- matrícula em posição intermediária;
- matrícula no último registro;
- matrícula inexistente.

Faça uma previsão antes de executar.

```python
testes = [
    ("primeiro", 1023),
    ("intermediário", 1082),
    ("último", 1091),
    ("inexistente", 9999),
]

for situacao, matricula in testes:
    resultado, comparacoes = buscar_contando(estudantes, matricula)
    print(f"{situacao:14s} | comparações={comparacoes} | resultado={resultado}")
```

## n = tamanho da entrada

Neste problema:

> **n = quantidade de estudantes na coleção**

A contagem de comparações permite discutir custo sem depender da velocidade específica do computador.

No melhor caso, podemos encontrar o estudante imediatamente. No pior caso, podemos precisar examinar toda a coleção.

## E quando os dados crescem?

Antes de executar qualquer experimento, preveja o número de comparações no **pior caso**:

| n | Comparações no pior caso |
|---:|---:|
| 10 | ? |
| 100 | ? |
| 1.000 | ? |
| 10.000 | ? |
| 1.000.000 | ? |

**O que ocorre quando `n` é multiplicado por 10?**

```python
def gerar_estudantes(n):
    return [
        {
            "matricula": 100000 + i,
            "nome": f"Estudante {i:06d}",
            "curso": "Ciência de Dados" if i % 3 else "Engenharia",
        }
        for i in range(n)
    ]

for n in [10, 100, 1_000, 10_000]:
    dados = gerar_estudantes(n)
    _, comparacoes = buscar_contando(dados, -1)  # matrícula inexistente = pior caso
    print(f"n={n:>6,d} | comparações={comparacoes:>6,d}")
```

## T(n): custo em função do tamanho da entrada

Podemos representar o custo do algoritmo como uma função do tamanho da entrada:

\[
T(n) = \text{quantidade de operações para uma entrada de tamanho } n
\]

Para uma relação linear:

\[
T(n) = a\cdot n + b
\]

Portanto:

\[
T(n) \propto n
\]

O custo cresce proporcionalmente ao tamanho da entrada.

## Big-O: ordem de crescimento

Nesta primeira semana, usaremos Big-O apenas para comunicar **ordem de crescimento**:

\[
a\cdot n+b \longrightarrow O(n)
\]

> **Big-O descreve uma ordem de crescimento do custo em função do tamanho da entrada.**

Não precisamos ainda da definição assintótica formal.

## Comparação qualitativa

Considere apenas, por enquanto:

\[
O(1), \qquad O(n), \qquad O(n^2)
\]

Se `n` dobrar, o que esperamos qualitativamente?

- `O(1)` → essencialmente não depende de `n`;
- `O(n)` → aproximadamente dobra;
- `O(n²)` → aproximadamente quadruplica.

O objetivo não é memorizar uma tabela, mas interpretar crescimento.

## E a nossa busca?

Retornemos ao problema concreto.

**Melhor caso:** `O(1)`  
**Pior caso:** `O(n)`

A pergunta importante é:

> **Por quê?**

Evite a frase isolada “busca sequencial é `O(n)`”. Sempre pergunte: **em qual caso?**

```python
def demonstrar_casos(n=10):
    dados = gerar_estudantes(n)
    casos = {
        "primeiro": dados[0]["matricula"],
        "meio": dados[n // 2]["matricula"],
        "último": dados[-1]["matricula"],
        "inexistente": -1,
    }

    for nome, matricula in casos.items():
        _, comparacoes = buscar_contando(dados, matricula)
        print(f"{nome:12s}: {comparacoes:>8,d} comparações")

demonstrar_casos(10)
```

## Big-O não é cronômetro

> **Tempo de execução observado ≠ ordem de crescimento**

Dois computadores podem obter tempos diferentes para a mesma tarefa.

O número de comparações determinado pela estrutura do algoritmo, para o mesmo caso e o mesmo `n`, não muda por causa da velocidade do processador.

```python
def medir_busca(dados, matricula, repeticoes=7):
    tempos = []
    for _ in range(repeticoes):
        inicio = perf_counter()
        buscar(dados, matricula)
        tempos.append(perf_counter() - inicio)
    return median(tempos)


dados = gerar_estudantes(100_000)
for i in range(5):
    tempo = medir_busca(dados, -1, repeticoes=1)
    print(f"execução {i + 1}: {tempo:.8f} s")
```

### Perguntas para discussão

- Por que os tempos podem variar entre execuções?
- Uma execução mais rápida transforma a busca sequencial em `O(1)`?
- Qual medida descreve melhor a relação estrutural entre tamanho da entrada e trabalho: segundos ou número de comparações?

## Vamos testar nossa hipótese

**Hipótese:**

> Se `n` aumentar, o número de comparações da busca sequencial crescerá linearmente no pior caso.

Entradas:

**1.000 → 10.000 → 100.000 → 1.000.000**

Antes de executar:

1. se `n` aumentar 10 vezes, o que esperamos para as comparações?
2. esperamos exatamente a mesma razão para o tempo?

```python
tamanhos = [1_000, 10_000, 100_000, 1_000_000]
resultados = []

for n in tamanhos:
    dados = gerar_estudantes(n)
    _, comparacoes = buscar_contando(dados, -1)
    tempo = medir_busca(dados, -1, repeticoes=5)
    resultados.append({"n": n, "comparacoes": comparacoes, "tempo_s": tempo})

for r in resultados:
    print(
        f"n={r['n']:>9,d} | "
        f"comparações={r['comparacoes']:>9,d} | "
        f"tempo mediano={r['tempo_s']:.8f} s"
    )
```
