## Uma primeira solução

<div class="three-words">
  <span class="fragment">Percorrer.</span>
  <span class="fragment blue">Comparar.</span>
  <span class="fragment gold">Encontrar.</span>
</div>

<div class="fragment prompt">Funciona. Então o problema está resolvido?</div>

```python
# Preparação: somente biblioteca padrão do Python.
from time import perf_counter
from statistics import median

estudantes = [
    {"matricula": 1023, "nome": "Ana", "curso": "Ciência de Dados"},
    {"matricula": 1047, "nome": "Bruno", "curso": "Ciência de Dados"},
    {"matricula": 1082, "nome": "Carla", "curso": "Engenharia"},
    {"matricula": 1091, "nome": "Diego", "curso": "Ciência de Dados"},
]

def buscar(estudantes, matricula):
    for estudante in estudantes:
        if estudante["matricula"] == matricula:
            return estudante
    return None

buscar(estudantes, 1082)
```

```python
def buscar_contando(estudantes, matricula):
    comparacoes = 0

    for estudante in estudantes:
        comparacoes += 1
        if estudante["matricula"] == matricula:
            return estudante, comparacoes

    return None, comparacoes

buscar_contando(estudantes, 1082)
```

```python
#| echo: false
def gerar_estudantes(n):
    return [
        {
            "matricula": 100000 + i,
            "nome": f"Estudante {i:06d}",
            "curso": "Ciência de Dados" if i % 3 else "Engenharia",
        }
        for i in range(n)
    ]

for n in [10, 100, 1_000, 10_000]:
    dados = gerar_estudantes(n)
    _, comparacoes = buscar_contando(dados, -1)
    print(f"n={n:>6,d} | comparações={comparacoes:>6,d}")
```

## <span class="big-o">O(n)</span> {.concept-slide}

<div class="concept-definition">ordem de crescimento</div>

<div class="fragment small-note">O custo cresce proporcionalmente ao tamanho da entrada.</div>

## Como diferentes custos crescem?

<div class="small-note">Se <strong>n</strong> dobrar, o que esperamos para cada ordem de crescimento?</div>

```python
import matplotlib.pyplot as plt

n = list(range(1, 11))
o1 = [1 for _ in n]
on = n
on2 = [x**2 for x in n]

AZUL = "#0094D4"
DOURADO = "#CCAC71"
CINZA = "#818285"

fig, ax = plt.subplots(figsize=(8.4, 4.8))
ax.plot(n, o1, linewidth=3, label="O(1)", color=CINZA)
ax.plot(n, on, linewidth=3, label="O(n)", color=AZUL)
ax.plot(n, on2, linewidth=3, label="O(n²)", color=DOURADO)
ax.set_xlabel("tamanho da entrada (n)")
ax.set_ylabel("crescimento relativo do custo")
ax.set_title("Comparação qualitativa de ordens de crescimento")
ax.legend(frameon=False)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(alpha=0.15)
plt.tight_layout()
plt.show()
```

```python
def demonstrar_casos(n=10):
    dados = gerar_estudantes(n)
    casos = {
        "primeiro": dados[0]["matricula"],
        "meio": dados[n // 2]["matricula"],
        "último": dados[-1]["matricula"],
        "inexistente": -1,
    }

    for nome, matricula in casos.items():
        _, comparacoes = buscar_contando(dados, matricula)
        print(f"{nome:12s}: {comparacoes:>8,d} comparações")

demonstrar_casos(10)
```

## Big-O não é cronômetro

<div class="context-line">Mesmo algoritmo. Mesmo <strong>n</strong>. Mesmo pior caso.</div>

::: {.fragment .question-big}
Os tempos de execução serão iguais?
:::

::: {.fragment .question-big}
E o número de comparações?
:::

::: {.fragment .inequality}
TEMPO OBSERVADO <span class="gold">≠</span> ORDEM DE CRESCIMENTO
:::

```python
def medir_busca(dados, matricula, repeticoes=7):
    tempos = []
    for _ in range(repeticoes):
        inicio = perf_counter()
        buscar(dados, matricula)
        tempos.append(perf_counter() - inicio)
    return median(tempos)

dados = gerar_estudantes(100_000)

for i in range(5):
    tempo = medir_busca(dados, -1, repeticoes=1)
    print(f"execução {i + 1}: {tempo:.8f} s")
```

## <span class="gold">HIPÓTESE</span>

<div class="hypothesis">
Se <strong>n</strong> aumentar, o número de comparações da busca sequencial crescerá
<strong>linearmente</strong> no pior caso.
</div>

<div class="fragment small-note">1.000 → 10.000 → 100.000 → 1.000.000</div>

```python
tamanhos = [1_000, 10_000, 100_000, 1_000_000]
resultados = []

for n in tamanhos:
    dados = gerar_estudantes(n)
    _, comparacoes = buscar_contando(dados, -1)
    tempo = medir_busca(dados, -1, repeticoes=5)

    resultados.append({
        "n": n,
        "comparacoes": comparacoes,
        "tempo_s": tempo
    })

for r in resultados:
    print(
        f"n={r['n']:>9,d} | "
        f"comparações={r['comparacoes']:>9,d} | "
        f"tempo mediano={r['tempo_s']:.8f} s"
    )
```

```python
# Visualização da evidência estrutural do experimento
import matplotlib.pyplot as plt

ns = [r["n"] for r in resultados]
comparacoes = [r["comparacoes"] for r in resultados]

fig, ax = plt.subplots(figsize=(8.4, 4.8))
ax.plot(ns, comparacoes, marker="o", linewidth=3, markersize=7, color="#0094D4")
ax.set_xlabel("tamanho da entrada (n)")
ax.set_ylabel("número de comparações")
ax.set_title("Busca sequencial — pior caso")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(alpha=0.15)
plt.tight_layout()
plt.show()
```
