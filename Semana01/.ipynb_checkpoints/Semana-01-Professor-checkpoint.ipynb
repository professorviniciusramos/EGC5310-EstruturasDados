## 0. Preparação

O notebook usa apenas a biblioteca padrão do Python. Isso reduz dependências e permite execução local ou no Google Colab.

A semente aleatória é fixada para tornar os experimentos reprodutíveis.


```python
from random import Random
from time import perf_counter
from statistics import median

RNG = Random(5310)
```

## 1. Busca sequencial

Começamos com uma coleção pequena de registros acadêmicos. A estrutura escolhida neste primeiro momento é deliberadamente simples: uma lista de dicionários.


```python
estudantes = [
    {"matricula": 1023, "nome": "Ana", "curso": "Ciência de Dados"},
    {"matricula": 1047, "nome": "Bruno", "curso": "Ciência de Dados"},
    {"matricula": 1082, "nome": "Carla", "curso": "Engenharia"},
    {"matricula": 1091, "nome": "Diego", "curso": "Ciência de Dados"},
]

estudantes
```

```python
def buscar(estudantes, matricula):
    for estudante in estudantes:
        if estudante["matricula"] == matricula:
            return estudante
    return None

buscar(estudantes, 1082)
```

### Perguntas para a turma

- A função resolve o problema?
- O que acontece se a matrícula estiver no primeiro registro?
- E se estiver no último?
- E se não existir?
- O fato de o algoritmo estar correto é suficiente para encerrarmos a análise?


## 2. Contando comparações

Em vez de começar pelo cronômetro, vamos observar diretamente uma operação relevante do algoritmo: **quantas matrículas são comparadas** até a busca terminar?


```python
def buscar_contando(estudantes, matricula):
    comparacoes = 0

    for estudante in estudantes:
        comparacoes += 1
        if estudante["matricula"] == matricula:
            return estudante, comparacoes

    return None, comparacoes
```

```python
testes = [
    ("primeiro", 1023),
    ("intermediário", 1082),
    ("último", 1091),
    ("inexistente", 9999),
]

for situacao, matricula in testes:
    resultado, comparacoes = buscar_contando(estudantes, matricula)
    print(f"{situacao:14s} | matrícula={matricula} | comparações={comparacoes} | resultado={resultado}")
```

### Ponto de discussão

Chamaremos de **n** o tamanho da entrada. Neste problema, `n` é a quantidade de estudantes na coleção.

A contagem de comparações nos permite discutir custo sem depender da velocidade específica do computador.


## 3. Melhor e pior caso

Vamos construir coleções maiores e controlar a posição do estudante procurado.


```python
def gerar_estudantes(n):
    return [
        {
            "matricula": 100000 + i,
            "nome": f"Estudante {i:06d}",
            "curso": "Ciência de Dados" if i % 3 else "Engenharia"
        }
        for i in range(n)
    ]

def demonstrar_casos(n=10):
    dados = gerar_estudantes(n)

    casos = {
        "primeiro": dados[0]["matricula"],
        "meio": dados[n // 2]["matricula"],
        "último": dados[-1]["matricula"],
        "inexistente": -1,
    }

    for nome, matricula in casos.items():
        _, comparacoes = buscar_contando(dados, matricula)
        print(f"{nome:12s}: {comparacoes:>8,d} comparações")

demonstrar_casos(10)
```

```python
for n in [10, 100, 1_000, 10_000]:
    dados = gerar_estudantes(n)
    _, melhor = buscar_contando(dados, dados[0]["matricula"])
    _, pior = buscar_contando(dados, -1)
    print(f"n={n:>6,d} | melhor caso={melhor:>2,d} | pior caso={pior:>6,d}")
```

### Interpretação

No melhor caso, encontramos o estudante na primeira comparação:

\[
T(n) \approx \text{constante} \Rightarrow O(1)
\]

No pior caso, examinamos todos os `n` estudantes:

\[
T(n) = a\cdot n + b
\]

Para entradas grandes, interessa principalmente a forma como o custo cresce com `n`:

\[
T(n) \propto n \Rightarrow O(n)
\]

Nesta primeira semana, a notação é usada para expressar **ordem de crescimento**, sem aprofundar ainda a definição assintótica formal.


## 4. Tempo × número de operações

Agora introduzimos o cronômetro. O objetivo não é substituir a análise por medições, mas mostrar por que **tempo observado** e **ordem de crescimento** são conceitos diferentes.


```python
def medir_busca(dados, matricula, repeticoes=7):
    tempos = []

    for _ in range(repeticoes):
        inicio = perf_counter()
        buscar(dados, matricula)
        tempos.append(perf_counter() - inicio)

    return median(tempos)

dados = gerar_estudantes(100_000)
matricula_inexistente = -1

for i in range(5):
    tempo = medir_busca(dados, matricula_inexistente, repeticoes=1)
    print(f"execução {i + 1}: {tempo:.8f} s")
```

### Perguntas para a turma

- Por que os tempos podem variar entre execuções?
- Uma execução mais rápida transforma a busca sequencial em `O(1)`?
- Qual medida descreve melhor a relação estrutural entre tamanho da entrada e trabalho: segundos ou número de comparações?


## 5. Experimento — crescimento da busca sequencial

**Hipótese:** se `n` aumentar, o número de comparações da busca sequencial crescerá linearmente no pior caso.

Usaremos matrículas inexistentes para garantir o pior caso. O tempo será medido pela mediana de várias execuções.


```python
tamanhos = [1_000, 10_000, 100_000, 1_000_000]
resultados = []

for n in tamanhos:
    dados = gerar_estudantes(n)

    _, comparacoes = buscar_contando(dados, -1)
    tempo = medir_busca(dados, -1, repeticoes=5)

    resultados.append({
        "n": n,
        "comparacoes": comparacoes,
        "tempo_s": tempo
    })

for r in resultados:
    print(
        f"n={r['n']:>9,d} | "
        f"comparações={r['comparacoes']:>9,d} | "
        f"tempo mediano={r['tempo_s']:.8f} s"
    )
```

```python
print("Razão de crescimento entre tamanhos consecutivos:\n")

for anterior, atual in zip(resultados, resultados[1:]):
    razao_n = atual["n"] / anterior["n"]
    razao_comp = atual["comparacoes"] / anterior["comparacoes"]
    razao_tempo = atual["tempo_s"] / anterior["tempo_s"]

    print(
        f"{anterior['n']:,} → {atual['n']:,}: "
        f"n ×{razao_n:.1f} | "
        f"comparações ×{razao_comp:.1f} | "
        f"tempo ×{razao_tempo:.2f}"
    )
```

### Leitura esperada

A razão entre as comparações acompanha exatamente a razão entre os tamanhos da entrada no pior caso. Já o tempo tende a crescer, mas não necessariamente na mesma proporção em cada execução.

Esse contraste sustenta a distinção:

> **benchmark mede uma execução em determinado ambiente; análise de complexidade descreve o padrão de crescimento do algoritmo.**


## 6. Visualização opcional do experimento

Esta célula é útil se houver tempo em sala. Ela usa `matplotlib`, normalmente disponível em Jupyter e Colab.


```python
import matplotlib.pyplot as plt

ns = [r["n"] for r in resultados]
comparacoes = [r["comparacoes"] for r in resultados]

plt.figure(figsize=(8, 4.5))
plt.plot(ns, comparacoes, marker="o")
plt.xlabel("Tamanho da entrada (n)")
plt.ylabel("Número de comparações")
plt.title("Busca sequencial no pior caso")
plt.grid(True, alpha=0.25)
plt.show()
```

## 7. Síntese e problema em aberto

Até aqui temos uma solução correta e simples:

1. os estudantes estão organizados em uma coleção;
2. percorremos a coleção sequencialmente;
3. no melhor caso fazemos uma comparação;
4. no pior caso fazemos `n` comparações;
5. o pior caso apresenta crescimento linear: `O(n)`.
