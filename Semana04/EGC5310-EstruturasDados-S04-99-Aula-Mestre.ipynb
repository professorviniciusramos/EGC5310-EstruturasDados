## Transformando decisões em variáveis

| Ideia | Variável |
|---|---|
| primeira posição possível | `inicio` |
| última posição possível | `fim` |
| posição examinada | `meio` |

```python
meio = (inicio + fim) // 2
```

> Depois da comparação, qual limite deve mudar?


## Implementação completa {.smaller}

As três decisões construídas coletivamente aparecem agora no código.


````python
def busca_binaria(matriculas, alvo):
    inicio = 0
    fim = len(matriculas) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        valor = matriculas[meio]

        if valor == alvo:
            return meio
        elif valor < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1

busca_binaria(matriculas, 84)
````

## O que garante o progresso?

- `inicio = meio + 1`
- `fim = meio - 1`

Usar apenas `meio` novamente pode manter o mesmo intervalo.

### E se o alvo não existir?

O intervalo termina quando:

```python
inicio > fim
```


## Contando e registrando o caminho {.smaller}

Esta versão torna o trabalho do algoritmo observável.


````python
def busca_binaria_detalhada(dados, alvo):
    inicio, fim = 0, len(dados) - 1
    comparacoes, caminho = 0, []
    while inicio <= fim:
        meio = (inicio + fim) // 2
        comparacoes += 1
        caminho.append((inicio, fim, meio, dados[meio]))
        if dados[meio] == alvo:
            return meio, comparacoes, caminho
        if dados[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1, comparacoes, caminho

busca_binaria_detalhada(matriculas, 70)
````

## A pré-condição não é detalhe

```python
[11, 18, 84, 31, 42, 57, 63, 71, 25, 92, 97]
```

> A busca binária ainda pode descartar uma metade com segurança?

**Não.** A coleção deve estar ordenada pela **mesma chave** usada na busca.


## Testando a pré-condição {.smaller}

O elemento existe, mas a coleção está desordenada.


````python
desordenadas = [11, 18, 84, 31, 42, 57, 63, 71, 25, 92, 97]
print("84 existe?", 84 in desordenadas)
print("resultado binário:", busca_binaria(desordenadas, 84))
````

## Correção depende de condições sobre os dados

- ordenado por **nome** → pode buscar binariamente por nome;
- ordenado por **matrícula** → pode buscar binariamente por matrícula;
- ordenado por nome e busca por matrícula → a pré-condição não está satisfeita.

> Um código pode estar correto e ainda produzir resposta inadequada quando sua pré-condição é violada.

````python
import math

for n in [10, 100, 1_000, 1_000_000, 1_000_000_000]:
    print(f"n={n:>13,} | sequencial ≤ {n:>13,} | binária ≤ {math.ceil(math.log2(n + 1)):>2}")
````

## Busca sequencial instrumentada {.smaller}

Primeiro contamos comparações; o cronômetro entra depois.


````python
import time
import statistics

def busca_sequencial_contada(dados, alvo):
    comparacoes = 0
    for i, valor in enumerate(dados):
        comparacoes += 1
        if valor == alvo:
            return i, comparacoes
    return -1, comparacoes
````

## Busca binária instrumentada {.smaller}

O contador aumenta uma vez por posição central examinada.


````python
def busca_binaria_contada(dados, alvo):
    inicio, fim, comparacoes = 0, len(dados) - 1, 0
    while inicio <= fim:
        meio = (inicio + fim) // 2
        comparacoes += 1
        if dados[meio] == alvo:
            return meio, comparacoes
        if dados[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1, comparacoes
````

## Execução do benchmark {.smaller}

Usaremos várias repetições e a mediana.


````python
def medir(funcao, dados, alvo, repeticoes=11):
    tempos = []
    resultado = None
    for _ in range(repeticoes):
        t0 = time.perf_counter_ns()
        resultado = funcao(dados, alvo)
        tempos.append(time.perf_counter_ns() - t0)
    return resultado, statistics.median(tempos)

for n in [100, 1_000, 10_000, 100_000, 1_000_000]:
    dados = list(range(n))
    alvo = -1
    seq, t_seq = medir(busca_sequencial_contada, dados, alvo)
    bin_, t_bin = medir(busca_binaria_contada, dados, alvo)
    print(n, seq[1], bin_[1], t_seq, t_bin)
````

## Como interpretar?

1. A contagem corresponde ao mecanismo previsto?
2. A diferença cresce com `n`?
3. O tempo é perfeitamente regular?
4. Ruído contradiz o Big-O?
5. Para entradas pequenas, constantes podem dominar?

> **Big-O explica a tendência; o cronômetro mede uma execução concreta.**
